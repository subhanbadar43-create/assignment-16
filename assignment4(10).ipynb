```python
def calculatehotelbill(roomtype, nights, breakfast=False):
    if roomtype == "Standard":
        roomprice = 5000
    elif roomtype == "Deluxe":
        roomprice = 8000
    elif roomtype == "Suite":
        roomprice = 15000
    else:
        roomprice = 0
        print("Invalid room type!")
    baseamount = roomprice * nights
    if nights >= 7:
        discountpercent = 15
    elif nights >= 3:
        discountpercent = 10
    else:
        discountpercent = 0
    discountamount = (baseamount * discountpercent) / 100
    if breakfast == True:
        breakfastcharge = 1000 * nights
    else:
        breakfastcharge = 0
    total = baseamount - discountamount + breakfastcharge
    print("Room Type =", roomtype)
    print("Nights =", nights)
    print("Room Charges =", baseamount)
    print("Discount (", discountpercent, "%) =", discountamount)
    print("Breakfast Charges =", breakfastcharge)
    print("Total Bill =", total)
    print("------------------------")
    return total
calculatehotelbill("Standard", 2)
calculatehotelbill("Deluxe", 5)
calculatehotelbill("Suite", 8)
calculatehotelbill("Deluxe", 5, breakfast=True)
```
